En un entorno real manejamos colecciones de indicadores de compromiso (IoCs) y tablas de servicios:
1. **Listas `[]`:** Colección ordenada y modificable. Añadimos elementos al final con `.append()`.
2. **Diccionarios `{clave: valor}`:** Guardan pares asociados para buscar un valor por su clave.
3. **Sets (Conjuntos) `set()`:** Colecciones que **eliminan automáticamente cualquier valor duplicado**.
