# 📌 MÓDULO 4: Networking Puro I - Direccionamiento IP y Subredes (`ipaddress`)
### 📖 Parte A: Aterrizando el Tema con un Ejemplo Guiado
El módulo estándar `ipaddress` permite calcular subredes CIDR sin hacer binario a mano:
* `ipaddress.ip_address("...")`: Valida si una IP es privada (`.is_private`) o pública (`.is_global`).
* `ipaddress.ip_network(".../CIDR")`: Calcula máscara (`.netmask`), broadcast (`.broadcast_address`) y hosts útiles (`.hosts()`).

**⏱️ Importante:** Al ejecutar esta celda de ejemplo se iniciará automáticamente el cronómetro para resolver el Caso Práctico 4.


````python
# ==============================================================================
# 📖 EJEMPLO GUIADO MÓDULO 4: DIRECCIONAMIENTO IP Y SUBREDES CON 'ipaddress'
# ==============================================================================
import ipaddress
import time
_inicio_m4 = time.time() # ⏱️ Inicia el cronómetro del Módulo 4

mi_ip = ipaddress.ip_address("192.168.10.45")
ip_externa = ipaddress.ip_address("8.8.8.8")

print(f"¿La IP {mi_ip} es privada (LAN)?: {mi_ip.is_private}")
print(f"¿La IP {ip_externa} es pública (WAN)?: {ip_externa.is_global}")

red_vlan_iot = ipaddress.ip_network("10.10.20.0/29")

print(f"\n--- Análisis de la Subred {red_vlan_iot} ---")
print("• Máscara de subred       :", red_vlan_iot.netmask)
print("• Dirección de Broadcast  :", red_vlan_iot.broadcast_address)
print("• Total de Hosts útiles   :", len(list(red_vlan_iot.hosts())))

ip_alerta = ipaddress.ip_address("10.10.20.5")
if ip_alerta in red_vlan_iot:
    print(f"⚠️ Confirmado: El host {ip_alerta} pertenece internamente a la VLAN IoT ({red_vlan_iot}).")

print("\n⏱️ [CRONÓMETRO MÓDULO 4 INICIADO] -> ¡Pasa a la siguiente celda y resuelve el Caso Práctico 4!")
````
