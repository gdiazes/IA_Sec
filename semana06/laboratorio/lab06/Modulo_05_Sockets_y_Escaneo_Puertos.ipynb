# 📌 MÓDULO 5: Networking Puro II - DNS, Sockets y Escáner de Puertos
### 📖 Parte A: Aterrizando el Tema con un Ejemplo Guiado
La librería `socket` nos permite resolver DNS (`gethostbyname`) y construir escáneres de puertos TCP (`SOCK_STREAM`) usando `.connect_ex((ip, puerto))`, que devuelve **`0`** cuando el puerto está **ABIERTO**.

**⏱️ Importante:** Al ejecutar esta celda de ejemplo se iniciará automáticamente el cronómetro para resolver el Caso Práctico 5.


````python
# ==============================================================================
# 📖 EJEMPLO GUIADO MÓDULO 5: RESOLUCIÓN DNS Y ESCÁNER DE PUERTOS CON 'socket'
# ==============================================================================
import socket
import time
_inicio_m5 = time.time() # ⏱️ Inicia el cronómetro del Módulo 5

ip_localhost = socket.gethostbyname("localhost")
print(f"🌐 Resolución DNS: 'localhost' apunta a la dirección IP -> {ip_localhost}\n")

servidor_demo = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
servidor_demo.bind(("127.0.0.1", 0))
servidor_demo.listen(1)
puerto_activo = servidor_demo.getsockname()[1]

puertos_a_probar = [puerto_activo, 9999]
print(f"Iniciando escaneo TCP sobre {ip_localhost}...")

for puerto in puertos_a_probar:
    cliente = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    cliente.settimeout(0.5)
    resultado = cliente.connect_ex((ip_localhost, puerto))

    if resultado == 0:
        print(f"  [🟢 ABIERTO] Puerto {puerto}/TCP (Código devuelto: {resultado})")
    else:
        print(f"  [🔴 CERRADO] Puerto {puerto}/TCP (Código devuelto: {resultado})")
    cliente.close()

servidor_demo.close()
print("\n⏱️ [CRONÓMETRO MÓDULO 5 INICIADO] -> ¡Pasa a la siguiente celda y resuelve el Caso Práctico 5!")
````
