# 📌 MÓDULO 1: Variables, Tipos de Datos y Manipulación de Strings (Logs y Tráfico)
### 📖 Parte A: Aterrizando el Tema con un Ejemplo Guiado
En ciberseguridad y redes, el 80% del tiempo procesamos texto (logs de firewalls, cabeceras HTTP, direcciones IP y MAC) y números (puertos, bytes, tiempos de respuesta TTL).
* **`str` (String):** Cadenas de texto entre comillas. Tienen métodos súper útiles como `.split()`, `.replace()` y `.upper()` / `.lower()`.
* **`int` y `float`:** Números enteros (puertos, bytes) y decimales (latencia, megabytes).
* **`bool`:** Valores lógicos (`True` o `False`).

**⏱️ Importante:** Al ejecutar esta celda de ejemplo se iniciará automáticamente el cronómetro para resolver el Caso Práctico 1.


````python
# ==============================================================================
# 📖 EJEMPLO GUIADO MÓDULO 1: VARIABLES, TIPOS Y DESTRIPADO DE UN LOG DE FIREWALL
# ==============================================================================
import time
_inicio_m1 = time.time() # ⏱️ Inicia el cronómetro del Módulo 1

ip_origen = "192.168.1.50"          # str (Texto: direcciones IP, MAC, dominios)
mac_address = "00:1a:2b:3c:4d:5e"   # str (Texto)
puerto_destino = 443                # int (Entero: puertos TCP/UDP, bytes)
latencia_ms = 14.75                 # float (Decimal: tiempos de respuesta, promedios)
puerto_abierto = True               # bool (Booleano: True / False)

print(f"[*] Analizando host {ip_origen} (MAC: {mac_address.upper()}) en puerto {puerto_destino}")
print(f"[*] Latencia: {latencia_ms} ms | ¿Puerto Abierto?: {puerto_abierto}\n")

raw_log = "2026-05-10 08:15:22 | DENY | TCP | SRC=10.0.0.25 | DST=172.16.0.10 | DPT=22"
print("Log original crudo:", raw_log)

partes = raw_log.split(" | ")
print("Log dividido en partes (índices 0 al 5):", partes)

fecha = partes[0]
accion = partes[1]
ip_atacante = partes[3].replace("SRC=", "")
puerto_atacado = int(partes[5].replace("DPT=", ""))

print(f"\n🚨 Resumen del SOC: El día {fecha}, la IP {ip_atacante} recibió un {accion} hacia el puerto {puerto_atacado}.")
print("\n⏱️ [CRONÓMETRO MÓDULO 1 INICIADO] -> ¡Pasa a la siguiente celda y resuelve el Caso Práctico 1!")
````
