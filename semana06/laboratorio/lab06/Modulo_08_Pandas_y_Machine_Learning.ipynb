### 📖 Parte A: Aterrizando el Tema con un Ejemplo Guiado
1. **Pandas (`pd.DataFrame`):** Estructura flujos de red en tablas y calcula métricas como el promedio con `.mean()`.
2. **Scikit-Learn (`RandomForestClassifier`):** Entrena un modelo de IA con **`.fit(X, y)`** y clasifica conexiones nuevas con **`.predict(nuevos_datos)`**.

**⏱️ Importante:** Al ejecutar esta celda de ejemplo se iniciará automáticamente el cronómetro para resolver el Caso Práctico 8.


````python
# ==============================================================================
# 📖 EJEMPLO GUIADO MÓDULO 8: PANDAS Y MACHINE LEARNING APLICADO A TRÁFICO DE RED
# ==============================================================================
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import time
_inicio_m8 = time.time() # ⏱️ Inicia el cronómetro del Módulo 8

df_entrenamiento = pd.DataFrame({
    "paquetes_seg":  [10, 15, 12, 20, 800, 950, 1200, 850],
    "bytes_paquete": [500, 600, 550, 480, 64, 60, 64, 58],
    "es_ataque":     [0, 0, 0, 0, 1, 1, 1, 1]
})

print("📊 Vista previa de la tabla de entrenamiento (Pandas):")
print(df_entrenamiento.head(6))
print(f"\nPromedio de bytes por paquete en la red: {df_entrenamiento['bytes_paquete'].mean():.2f} bytes")

X_train = df_entrenamiento[["paquetes_seg", "bytes_paquete"]]
y_train = df_entrenamiento["es_ataque"]

modelo_ia = RandomForestClassifier(random_state=42)
modelo_ia.fit(X_train, y_train)

nuevo_flujo = pd.DataFrame({"paquetes_seg": [900], "bytes_paquete": [64]})
prediccion = modelo_ia.predict(nuevo_flujo)[0]
print(f"\n🤖 Veredicto de la IA para el nuevo flujo (1=Ataque, 0=Normal): {prediccion}")
print("\n⏱️ [CRONÓMETRO MÓDULO 8 INICIADO] -> ¡Pasa a la siguiente celda y resuelve el Caso Práctico 8!")
````
