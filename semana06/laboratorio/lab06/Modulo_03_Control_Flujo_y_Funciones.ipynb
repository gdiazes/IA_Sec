# 📌 MÓDULO 3: Control de Flujo, Bucles y Funciones (Reglas de Firewall / IDS)
### 📖 Parte A: Aterrizando el Tema con un Ejemplo Guiado
Un Firewall o IDS combina tres elementos fundamentales:
1. **Condicionales (`if / elif / else`):** Toman decisiones según condiciones.
2. **Funciones (`def`):** Empaquetan esa lógica en un bloque reutilizable (`return`).
3. **Bucles (`for`):** Recorren múltiples paquetes aplicando la función a cada uno.

**⏱️ Importante:** Al ejecutar esta celda de ejemplo se iniciará automáticamente el cronómetro para resolver el Caso Práctico 3.


````python
# ==============================================================================
# 📖 EJEMPLO GUIADO MÓDULO 3: MOTOR DE REGLAS DE UN FIREWALL CON DEF, FOR E IF
# ==============================================================================
import time
_inicio_m3 = time.time() # ⏱️ Inicia el cronómetro del Módulo 3

def evaluar_paquete(ip, puerto, intentos):
    ips_bloqueadas = ["185.220.101.5", "45.33.32.156"]
    puertos_prohibidos = [23, 445, 3389]

    if ip in ips_bloqueadas:
        return "🔴 DROP (IP en Lista Negra de Threat Intelligence)"
    elif puerto in puertos_prohibidos:
        return "🟠 REJECT (Puerto administrativo prohibido)"
    elif intentos > 10:
        return "🟡 ALERT (Posible Fuerza Bruta detectada)"
    else:
        return "🟢 ALLOW (Tráfico permitido)"

trafico_red = [
    {"ip": "192.168.1.20", "puerto": 443, "intentos": 1},
    {"ip": "185.220.101.5", "puerto": 80,  "intentos": 2},
    {"ip": "10.0.0.55",    "puerto": 445, "intentos": 1},
    {"ip": "172.16.0.9",   "puerto": 22,  "intentos": 25}
]

print("--- MOTOR DE FIREWALL EN EJECUCIÓN ---")
for paquete in trafico_red:
    decision = evaluar_paquete(paquete["ip"], paquete["puerto"], paquete["intentos"])
    print(f"Origen: {paquete['ip']:15} | Puerto: {paquete['puerto']:4} -> {decision}")

print("\n⏱️ [CRONÓMETRO MÓDULO 3 INICIADO] -> ¡Pasa a la siguiente celda y resuelve el Caso Práctico 3!")
````
