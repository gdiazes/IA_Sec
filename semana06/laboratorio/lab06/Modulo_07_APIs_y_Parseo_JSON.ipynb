# 📌 MÓDULO 7: Procesamiento de Respuestas JSON y APIs de Threat Intelligence (`json`)
### 📖 Parte A: Aterrizando el Tema con un Ejemplo Guiado
Las herramientas de ciberseguridad (SIEM, SOAR, VirusTotal, AbuseIPDB) se comunican enviando texto en formato **JSON**. Con **`json.loads(texto)`** convertimos ese texto inmediatamente en un diccionario de Python.

**⏱️ Importante:** Al ejecutar esta celda de ejemplo se iniciará automáticamente el cronómetro para resolver el Caso Práctico 7.


````python
# ==============================================================================
# 📖 EJEMPLO GUIADO MÓDULO 7: CONSUMO DE APIS Y PARSEO JSON DE THREAT INTEL
# ==============================================================================
import json
import time
_inicio_m7 = time.time() # ⏱️ Inicia el cronómetro del Módulo 7

respuesta_api_raw = '{"ip": "185.220.101.45", "pais": "DE", "isp": "Tor Exit Node", "reportes_previos": 142, "confianza_abuso": 98}'
print("Texto JSON crudo recibido de la API:", respuesta_api_raw)

datos_ti = json.loads(respuesta_api_raw)

ip_consultada = datos_ti["ip"]
isp_origen = datos_ti["isp"]
score_abuso = datos_ti["confianza_abuso"]

print("\n--- 🌍 REPORTE ENRIQUECIDO DEL SOC ---")
print(f"• IP Investigada      : {ip_consultada} ({datos_ti['pais']})")
print(f"• Proveedor (ISP)     : {isp_origen}")
print(f"• Nivel de Riesgo CTI : {score_abuso}% ({datos_ti['reportes_previos']} reportes)")
print("\n⏱️ [CRONÓMETRO MÓDULO 7 INICIADO] -> ¡Pasa a la siguiente celda y resuelve el Caso Práctico 7!")
````
